# Get the real-time daily data

Currently only the real-time daily data is supported, using the function realtimeUpdate(source, delta=60). Among them, source supports 'qq' | 'qmt', and qq is used by default.

**When using qq, you should control the interval between two data acquisitions (using the delta parameter, defaulting to 60s), so as not to put the ip on the blacklist of sina or qq.**
